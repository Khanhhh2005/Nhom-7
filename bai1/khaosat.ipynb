```python
# ============================================================
# BÀI 1 - HIỂU DỮ LIỆU, CHỌN DỮ LIỆU VÀ TIỀN XỬ LÝ
# DATASET 1: RETAIL SALES & CUSTOMER BEHAVIOR
# ============================================================

# 1. Import thư viện
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Pandas:", pd.__version__)
print("NumPy:", np.__version__)
print("Đã tải thư viện thành công!")


# 2. Đọc Dataset 1
retail_path = "../data/raw/archive/retail_data.csv"

retail = pd.read_csv(retail_path)


# 3. Kiểm tra quy mô dữ liệu
print("\n=== QUY MÔ DATASET 1 ===")
print("Kích thước dataset:", retail.shape)
print("Số bản ghi:", retail.shape[0])
print("Số thuộc tính:", retail.shape[1])


# 4. Hiển thị 5 bản ghi đầu tiên
print("\n=== 5 BẢN GHI ĐẦU TIÊN ===")
display(retail.head())


# 5. Kiểm tra cấu trúc dữ liệu
print("\n=== THÔNG TIN DATASET 1 ===")
retail.info()


# 6. Danh sách các thuộc tính
print("\n=== DANH SÁCH THUỘC TÍNH ===")
for i, col in enumerate(retail.columns, 1):
    print(f"{i}. {col}")

# ============================================================
# 7. KHẢO SÁT CHẤT LƯỢNG DỮ LIỆU - DATASET 1
# ============================================================

# --- Giá trị thiếu ---
missing_count = retail.isnull().sum()
missing_percent = (missing_count / len(retail) * 100).round(2)

missing_table = pd.DataFrame({
    "Số giá trị thiếu": missing_count,
    "Tỷ lệ thiếu (%)": missing_percent
})

missing_table = missing_table[missing_table["Số giá trị thiếu"] > 0]
missing_table = missing_table.sort_values(
    "Tỷ lệ thiếu (%)",
    ascending=False
)

print("=== GIÁ TRỊ THIẾU ===")

if len(missing_table) == 0:
    print("Không phát hiện giá trị thiếu.")
else:
    display(missing_table)


# --- Dữ liệu trùng lặp ---
duplicate_count = retail.duplicated().sum()

print("\n=== DỮ LIỆU TRÙNG LẶP ===")
print("Số dòng trùng lặp:", duplicate_count)
print(
    "Tỷ lệ trùng lặp:",
    round(duplicate_count / len(retail) * 100, 2),
    "%"
)


# --- Phân loại kiểu thuộc tính ---
numeric_cols = retail.select_dtypes(include=np.number).columns.tolist()
categorical_cols = retail.select_dtypes(exclude=np.number).columns.tolist()

print("\n=== KIỂU THUỘC TÍNH ===")
print("Số thuộc tính dạng số:", len(numeric_cols))
print("Số thuộc tính dạng phân loại/text:", len(categorical_cols))

print("\nThuộc tính dạng số:")
print(numeric_cols)

print("\nThuộc tính dạng phân loại/text:")
print(categorical_cols)


# --- Thống kê mô tả ---
print("\n=== THỐNG KÊ MÔ TẢ CÁC THUỘC TÍNH SỐ ===")
display(retail[numeric_cols].describe().T)
```

```python
# KIỂM TRA GIÁ TRỊ THIẾU VÀ TRÙNG LẶP

print("=== GIÁ TRỊ THIẾU ===")

missing = retail.isnull().sum()
missing = missing[missing > 0]

if len(missing) == 0:
    print("Không phát hiện giá trị thiếu.")
else:
    print(missing)

print("\n=== DỮ LIỆU TRÙNG LẶP ===")

duplicate_count = retail.duplicated().sum()

print("Số dòng trùng lặp:", duplicate_count)
print(
    "Tỷ lệ trùng lặp:",
    round(duplicate_count / len(retail) * 100, 2),
    "%"
)
```

```python
# ============================================================
# KIỂM TRA NGOẠI LAI (OUTLIER) - DATASET 1
# Phương pháp IQR
# ============================================================

# Lấy các thuộc tính dạng số
numeric_cols = retail.select_dtypes(include=np.number).columns

outlier_results = []

for col in numeric_cols:
    Q1 = retail[col].quantile(0.25)
    Q3 = retail[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (retail[col] < lower_bound) |
        (retail[col] > upper_bound)
    ).sum()

    outlier_percent = round(
        outlier_count / len(retail) * 100, 2
    )

    outlier_results.append([
        col,
        outlier_count,
        outlier_percent
    ])

outlier_table = pd.DataFrame(
    outlier_results,
    columns=[
        "Thuộc tính",
        "Số ngoại lai",
        "Tỷ lệ (%)"
    ]
)

# Chỉ hiển thị những thuộc tính có ngoại lai
outlier_table = outlier_table[
    outlier_table["Số ngoại lai"] > 0
].sort_values("Tỷ lệ (%)", ascending=False)

print("=== NGOẠI LAI THEO PHƯƠNG PHÁP IQR ===")

if len(outlier_table) == 0:
    print("Không phát hiện ngoại lai bằng phương pháp IQR.")
else:
    display(outlier_table)
```

## Đề xuất kỹ thuật khai phá dữ liệu

Sau khi khảo sát ba bộ dữ liệu, các kỹ thuật khai phá được đề xuất như sau:

| Dataset | Classification | Association | Clustering |
|---|---|---|---|
| Retail Sales & Customer Behavior | Có | Có | Có |
| Diabetes 130-US Hospitals | Có | Không | Có |
| Human Activity Recognition | Có | Không | Có |

### Dataset 1 - Retail Sales & Customer Behavior
- **Association:** Phân tích các sản phẩm hoặc nhóm sản phẩm thường xuất hiện cùng nhau trong giao dịch.
- **Classification:** Có thể sử dụng thuộc tính `churned` làm nhãn để xây dựng bài toán phân loại khách hàng.
- **Clustering:** Phân nhóm khách hàng dựa trên đặc điểm nhân khẩu học và hành vi mua sắm.

### Dataset 2 - Diabetes 130-US Hospitals
- **Classification:** Sử dụng thuộc tính `readmitted` làm nhãn để xây dựng bài toán phân loại tình trạng tái nhập viện.
- **Clustering:** Phân nhóm các lượt điều trị/bệnh nhân có đặc điểm tương đồng.

### Dataset 3 - Human Activity Recognition Using Smartphones
- **Classification:** Sử dụng `activity` làm nhãn để phân loại hoạt động của người dùng từ dữ liệu cảm biến.
- **Clustering:** Phân nhóm các mẫu dữ liệu cảm biến dựa trên mức độ tương đồng giữa các đặc trưng.

```python
# ============================================================
# TIỀN XỬ LÝ DATASET 1 - RETAIL
# ============================================================

# Tạo bản sao để không làm thay đổi dữ liệu gốc
retail_processed = retail.copy()

print("Kích thước trước tiền xử lý:", retail_processed.shape)


# 1. Chuyển transaction_date từ text sang datetime
retail_processed["transaction_date"] = pd.to_datetime(
    retail_processed["transaction_date"],
    errors="coerce"
)

print("\n=== KIỂM TRA TRANSACTION_DATE ===")
print("Kiểu dữ liệu:", retail_processed["transaction_date"].dtype)
print(
    "Số ngày không chuyển đổi được:",
    retail_processed["transaction_date"].isna().sum()
)


# 2. Tạo thuộc tính mới từ ngày giao dịch
retail_processed["transaction_year"] = (
    retail_processed["transaction_date"].dt.year
)

retail_processed["transaction_month"] = (
    retail_processed["transaction_date"].dt.month
)


# 3. Tạo giá trị giao dịch trước giảm giá
retail_processed["gross_value"] = (
    retail_processed["quantity"] *
    retail_processed["unit_price"]
)


# 4. Kiểm tra kết quả
print("\n=== THUỘC TÍNH MỚI ===")

display(
    retail_processed[
        [
            "transaction_date",
            "transaction_year",
            "transaction_month",
            "quantity",
            "unit_price",
            "gross_value"
        ]
    ].head()
)

print("\nKích thước sau khi thêm thuộc tính:", retail_processed.shape)
```

```python
# ============================================================
# KIỂM TRA THUỘC TÍNH THỜI GIAN TRÙNG THÔNG TIN
# ============================================================

month_match = (
    retail_processed["transaction_month"]
    == retail_processed["month_of_year"]
).mean() * 100

print("Tỷ lệ transaction_month trùng month_of_year:",
      round(month_match, 2), "%")

print("\nCác năm có trong dữ liệu:")
print(
    retail_processed["transaction_year"]
    .value_counts()
    .sort_index()
)

print("\nKiểm tra gross_value:")
print(retail_processed["gross_value"].describe())
```

```python
# ============================================================
# KIỂM TRA TÍNH NHẤT QUÁN CỦA THUỘC TÍNH THỜI GIAN
# ============================================================

# Tạo tháng trực tiếp từ transaction_date
month_from_date = retail_processed["transaction_date"].dt.month

# So sánh với month_of_year có sẵn
inconsistent_month = (
    month_from_date != retail_processed["month_of_year"]
)

inconsistent_count = inconsistent_month.sum()
inconsistent_percent = (
    inconsistent_count / len(retail_processed) * 100
)

print("=== KIỂM TRA TÍNH NHẤT QUÁN ===")
print("Số bản ghi month_of_year không khớp transaction_date:",
      inconsistent_count)

print("Tỷ lệ không khớp:",
      round(inconsistent_percent, 2), "%")

print("\nVí dụ các bản ghi không khớp:")

display(
    retail_processed.loc[
        inconsistent_month,
        [
            "transaction_date",
            "month_of_year",
            "transaction_month"
        ]
    ].head(10)
)
```

```python
# ============================================================
# HOÀN THIỆN TIỀN XỬ LÝ DATASET 1 - RETAIL
# ============================================================

# 1. Sửa month_of_year dựa trên transaction_date
retail_processed["month_of_year"] = (
    retail_processed["transaction_date"].dt.month
)

# transaction_month được tạo trước đó giờ đã dư thừa
retail_processed.drop(
    columns=["transaction_month"],
    inplace=True
)

# 2. Kiểm tra lại tính nhất quán
month_check = (
    retail_processed["month_of_year"]
    == retail_processed["transaction_date"].dt.month
).mean() * 100

print("=== KIỂM TRA SAU TIỀN XỬ LÝ ===")
print(
    "Tỷ lệ month_of_year khớp transaction_date:",
    round(month_check, 2),
    "%"
)

print("Số giá trị thiếu:",
      retail_processed.isnull().sum().sum())

print("Số dòng trùng lặp:",
      retail_processed.duplicated().sum())

print("Kích thước dữ liệu cuối cùng:",
      retail_processed.shape)
```
